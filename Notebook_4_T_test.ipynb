```python
import numpy as np
import pandas as pd
from scipy import stats
from IPython.display import display

pd.set_option("display.max_colwidth", 20)
pd.set_option("display.width", 200)

def style_table(df):
    def highlight(val):
        try:
            if val == "<0.001":
                return "background-color: #f7ecd8"
            if float(val) < 0.05:
                return "background-color: #f7ecd8"
        except:
            pass
        return ""

    return (
        df.style
        .applymap(highlight)
        .set_properties(**{
            "text-align": "center",
            "border": "1px solid #d9d9d9",
            "font-size": "11pt",
            "padding": "6px"
        })
        .set_table_styles([
            {"selector": "th",
             "props": [("background-color", "#f0f0f0"),
                       ("font-weight", "bold"),
                       ("border", "1px solid #d9d9d9"),
                       ("text-align", "center")]},
            {"selector": "caption",
             "props": [("caption-side", "bottom"),
                       ("font-size", "11pt"),
                       ("font-style", "italic"),
                       ("padding-top", "8px")]}
        ])
    )

def ttest_table(data, title, caption):
    df = pd.DataFrame(data)
    models = list(df.columns)
    n = len(models)

    means = df.mean()
    stds = df.std()

    rows = [f"{means[m]:.3f} +/- {stds[m]:.3f}" for m in models]
    cols = [chr(65+i) for i in range(n)]

    matrix = np.full((n, n), "", dtype=object)

    for i in range(n):
        for j in range(i+1, n):
            _, p = stats.ttest_rel(df[models[i]], df[models[j]])
            matrix[i, j] = "<0.001" if p < 0.001 else f"{p:.3f}"

    df_final = pd.DataFrame(matrix, index=rows, columns=cols)

    print("\n" + title)
    display(style_table(df_final).set_caption(caption))

accuracy_trained_accuracy = {
    "NN_2HL": [98.10,98.60,97.70,98.20,97.90,97.60,98.10,97.70,98.90,96.40],
    "Rule":  [98.30,97.70,98.00,98.40,98.80,97.10,98.40,97.80,98.80,98.00],
    "BayesNet": [83.00,83.80,83.80,83.80,83.30,81.40,83.90,83.60,83.50,83.60],
    "DecisionTree": [94.75,94.38,95.88,95.00,95.50,93.38,96.25,95.38,95.38,94.88],
    "NaiveBayes": [94.60,92.00,93.60,94.70,94.60,93.60,95.70,95.30,94.00,94.10],
    "SVM": [96.88,97.00,97.13,97.00,97.50,95.63,97.75,97.25,96.88,96.75]
}

ttest_table(
    accuracy_trained_accuracy,
    "1) T-Test on accuracy-trained models (accuracy)",
    "Table 7.1. T-Test on accuracy performance of accuracy-trained models"
)

accuracy_trained_cost = {
    "NN_2HL": [0.1523]*10,
    "Rule": [0.063,0.074,0.056,0.046,0.033,0.090,0.052,0.076,0.051,0.077],
    "BayesNet": [0.172,0.203,0.167,0.132,0.141,0.255,0.191,0.256,0.135,0.174],
    "DecisionTree": [0.091,0.137,0.239,0.200,0.174,0.190,0.179,0.159,0.178,0.161]
}

ttest_table(
    accuracy_trained_cost,
    "2) T-Test on accuracy-trained models (cost)",
    "Table 7.2. T-Test on cost performance of accuracy-trained models"
)

cost_trained_accuracy = {
    "NN_3HL": [97.70,97.70,95.40,96.00,96.50,97.60,95.90,95.20,96.70,96.20],
    "Rule": [95.50,95.30,95.30,95.40,96.70,94.60,95.70,94.20,94.90,95.00],
    "BayesNet": [94.50,95.00,95.90,95.80,96.70,94.30,96.20,94.20,96.40,95.20],
    "NaiveBayes": [68.50,68.00,69.10,69.30,70.50,71.00,71.10,72.00,70.30,70.20]
}

ttest_table(
    cost_trained_accuracy,
    "3) T-Test on cost-trained models (accuracy)",
    "Table 7.3. T-Test on accuracy performance of cost-trained models"
)

cost_trained_cost = {
    "NN_3HL": [0.059,0.059,0.100,0.067,0.053,0.033,0.059,0.093,0.042,0.110],
    "Rule": [0.063,0.074,0.056,0.046,0.033,0.090,0.052,0.076,0.051,0.077],
    "BayesNet": [0.172,0.203,0.167,0.132,0.141,0.255,0.191,0.256,0.135,0.174],
    "NaiveBayes": [0.147,0.173,0.122,0.132,0.158,0.137,0.126,0.138,0.092,0.122]
}

ttest_table(
    cost_trained_cost,
    "4) T-Test on cost-trained models (cost)",
    "Table 7.4. T-Test on cost performance of cost-trained models"
)
```
